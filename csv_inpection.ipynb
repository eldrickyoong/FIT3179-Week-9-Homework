```python
import pandas as pd
csv_path = './data/population_state.csv'
df = pd.read_csv(csv_path)
```

```python
df = df[
  (df['year'] == 2025) &
  (df['age'] == 'overall') &
  (df['ethnicity'] == 'overall') &
  (df['sex'] == 'both')
        ]
df
```

```python
df['state'] = df['state'].replace('^W.P. | ','', regex=True)
df = df.drop(columns=['age', 'date', 'sex', 'ethnicity'], errors='ignore')
```

```python
df.to_csv('./data/population_state_simple.csv', index=False)
```
